# 1 · Simple moving-average crossover strategy
Cleaned-up version of the original *Build a simple trading strategy* notebook.

**Rule:** hold one share while MA10 > MA50, otherwise stay in cash. Daily profit = tomorrow's close − today's close while holding.

Fixes vs. the original: `pd.DataFrame.from_csv` (removed from pandas) → `yfinance`/`read_csv`; `Close1` was used before it was created; `wealth` vs `Wealth` name mismatch; list comprehensions → vectorised pandas.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd, numpy as np, matplotlib.pyplot as plt
import config
from src import data_loader as dl
OFFLINE = False   # True = synthetic sample data (no internet needed)
```

```python
from src.strategy import ma_crossover, summary
data = dl.generate_sample_data(['MSFT']) if OFFLINE else dl.fetch_prices(['MSFT'], '2015-01-01', config.END_DATE)
bt = ma_crossover(data['MSFT'], fast=10, slow=50)
bt.tail()
```

```python
ax = bt[['Close', 'MA10', 'MA50']].plot(figsize=(12, 5), title='MSFT close with MA10 / MA50')
```

```python
ax = bt['Profit'].plot(figsize=(12, 4), title='Daily profit / loss ($ per share)')
ax.axhline(0, color='red', lw=1)
```

```python
ax = bt[['Wealth', 'BuyHold']].plot(figsize=(12, 5))
ax.set_title(f"Strategy profit ${bt['Wealth'].iloc[-1]:.2f}/share vs buy & hold ${bt['BuyHold'].iloc[-1]:.2f}/share")
summary(bt)
```
